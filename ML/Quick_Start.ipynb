## Step 2: Import ML Engine

```python
import sys
sys.path.append('..')

from ML import MLEngine, MLTrainer
print("✅ Imports successful!")
```

## Step 3: Train a Model (First Time Only)

```python
# Train model for SPY (S&P 500)
# This takes 2-5 minutes

trainer = MLTrainer('SPY', lookback_days=730)
metrics = trainer.full_training_pipeline(
    test_size=0.2,
    train_lstm=False,
    save_models=True
)

print(f"\n✅ Training complete!")
print(f"Accuracy: {metrics['ensemble']['accuracy']:.2%}")
```

## Step 4: Get Predictions

```python
# Initialize engine
engine = MLEngine()

# Get prediction
prediction = engine.predict('SPY')

# Display result
print(f"\n🎯 Prediction: {prediction['final_prediction']}")
print(f"🎲 Confidence: {prediction['final_confidence']:.2%}")
print(f"⚠️  Risk Level: {prediction['risk_level']}")

if 'probability' in prediction:
    print(f"\n📊 Probabilities:")
    print(f"   UP:   {prediction['probability']['up']:.2%}")
    print(f"   DOWN: {prediction['probability']['down']:.2%}")
```

## Step 5: Batch Predictions

```python
import pandas as pd

# Predict for multiple symbols
symbols = ['SPY', 'QQQ', 'DIA']
predictions = engine.batch_predict(symbols)

# Create summary
summary = []
for symbol, pred in predictions.items():
    if 'error' not in pred:
        summary.append({
            'Symbol': symbol,
            'Direction': pred['final_prediction'],
            'Confidence': f"{pred['final_confidence']:.2%}",
            'Risk': pred['risk_level']
        })

df = pd.DataFrame(summary)
display(df)
```

## Step 6: Market Regime Detection

```python
# Check market regime
regime = engine.get_market_regime('SPY')

emoji = '🐂' if regime == 'BULL' else '🐻' if regime == 'BEAR' else '↔️'
print(f"{emoji} Market Regime: {regime}")
```
